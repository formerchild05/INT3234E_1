```python
missing = fundpd.isna().sum()
missing_pct = fundpd.isna().mean() * 100

result = pd.DataFrame({
    'missing': missing,
    'missing_pct': missing_pct
})

print(result[result['missing'] > 0].sort_values('missing', ascending=False))
```
